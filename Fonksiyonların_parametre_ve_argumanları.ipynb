    # çıktı demet olarak verilmiştir.
# Verilen bütün değerleri çarpan program:

```python
def çarp(*args):  # * işareti parmetreyi ögelerine ayırıyor.
     sonuç=1
     for i in args: # pythonda * işareti ile birlikte kullanılan parametrenin adı geleneksel olarak 'args'dır.          
        sonuç *= i
     print(sonuç)
çarp(1,2,8,79854,5)
```

#  Rasgele Sayıda İsimli Parametre Belirleme

```python
def karşılık_bul(*args, **kwargs): # ** işareti fanksiyonu çağırırken yazdığımız isimli parametrelerin bize sözlük olarak verilmesini sağlıyor.
    for sözcük in args:
        if sözcük in kwargs:
            print("{}={}".format(sözcük,kwargs[sözcük]))
        else:
            print("{} kelimesi sözlükte yok!".format(sözcük))

sözlük={"kitap":"book",
        "bilgisayar":"computer",
        "programlama":"programmming"}
karşılık_bul("kitap","bilgisayar","fonksiyon",**sözlük)
```

    # burada tanımlanan karşılı_bul() isimli fonksiyon, kendisine veriken parametreleri(*args),
    # bir sözlük içinde arayarak (**sözlük) karşılıklarını çıktı olarak veriyor.
    # *args ifadesi ile, kullanıcıya istediği sayıda isimsiz parametre 
    # ,**kwargs ifadesiyle ise kullanıcıya istediği sayıda isimli parametre girme olanagı veriyoruz.

```python
def bas(*args ,start='', **kwargs): # bas adlı bir fonksiyon tanımladık.
     for öğe in args:
          print(start+öğe,**kwargs)
bas('öğe1','öğe2','öğe3',start="->")
```

# ÖRNEK PROGRAM
#Belli miktarda ve belli aralıkla sayı üreten program.

```python
import random # random modülünü içe aktardık.

def sayı_üret(başlangıç=0, bitiş=500, adet=6): # sayı_üret adlı bir fonksiyon tanımladık. bu fonsiyon üç farklı parametre alıyor.
    sayılar=set() # küme oluşturduk. Kümelerin içindeki öğeler her zaman tektir. yani aynı öğeden sadece bir tane bulunabilir.
    while len(sayılar) < adet: # sayılar değişkeninin uzunluğu adet parametresinin değerinden az olduğu sürece,
        sayılar.add(random.randrange(başlangıç,bitiş))# sayılar adlı değişkene baslangıç ve bitiş parametreleri arasından rastgele değerler ekleyeceğiz.
    return sayılar #  sayılar değişkenini döndürüyoruz.
print(sayı_üret())
```

# sample() fonksiyonu herhangi bir dizi içinden istenen sayıda numune alınmasını sağlar.

```python
liste=["ahmet","mehmet","hatice","sevim","sevinç","selim","zeynep","sevgi"]
print(random.sample(liste,2)) # listeden iki tane rastgele numune aldık.
```

# shuffle() adlı başka bir fonksiyon,  bir dizi içindeki öğelerin sırasını rastgele şekilde karıştırır.

```python
random.shuffle(liste)
print(liste)
```

# FONKSİYONLARIN KAPSAMI VE GLOBAL DEYİMİ
    # Fonksiyon içindeki ve dışındaki aynı isimli değişkenler birbirlerine karışmıyor.
    # Bunun sebebi Python'da değişkenlerin, fanksiyonların ve sınıfların bir kapsamı vardır. Bu kapsama 'isim alanı' adı verilir.

```python
x=0 #global alandaki x

def fonk():
    x=1 # lokal alandaki x
    print(x)

fonk()
print(x)
```

#python önce mevcut isim alanını kontrol ediyor. eğer değişken mevcut isim  alanında bulunuyorsa global alana bakmıyor.
#Ama eğer değişken mevcut isim alanında yoksa global alana bakıyor.
#Değiştirilebilir nesnelerin değerlerini, lokal isim alanlarından değiştirebilirsin:

```python
a=set()

def fnk():
    a.add(10)
    return a

print(fnk())
```


#bu sorunu global deyimi ile çözebiliriz. Ama elden geldiğince az kullanılmalı.

```python
isim="Bilge "

def fonk():
    global isim
    isim+="Barbar"
    return isim

print(fonk())
```
