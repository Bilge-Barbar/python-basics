```python
sesli_harfler = 'aeıioöuü'
saya = 0

kelime3 = input('Bir kelime girin: ')

def seslıdır(harfi):
   return harfi in sesli_harfler

def artır():
    global saya
    for harfi in kelime3:
       if seslıdır(harfi):
          saya +=1
    return saya

mesaj = '{} kelimesinde {} sesli harf var.'
print(mesaj.format(kelime3, artır()))
```

# ÖRNEK NİTELİKLERİ (İNSTANCE ATTRİBUTES)

# __init__ fonksiyonu ve self
 Bildiğiniz gibi, Python’da sınıf niteliklerini tanımlamak için yapmamız gereken tek şey, sınıf tanımının hemen altına bunları alelade birer değişken gibi yazmaktan ibarettir.
  Örnek niteliklerini tanımlamak için ise iki yardımcı araca ihtiyacımız var: __init__() fonksiyonu ve self.
   Bu özel fonksiyonun görevi, sınıfımızı örneklediğimiz sırada, yani mesela ahmet = Çalışan() gibi bir komut verdiğimiz anda oluşturulacak nitelikleri ve gerçekleştirilecek işlemleri tanımlamaktır.
self kelimesi, Python programlama dilinin söz diziminin
gerektirdiği bir öğedir. Bu kelime, Çalışan() adlı sınıfın örneklerini temsil eder.Fonksiyon dışında örnek niteliği tanımlayamazsınız.(__init__ in altında olmalı).
örnek çıkarırkenkş genel yapı: nesne_adi = SınıfAdı(parametreler)

```python
class Çalışan(): #sınıf tanımlandı
    def __init__(self): #kurucu metot
        self.kabiliyetleri=[] #Çalışan nesnesinin bir özelliği
        print(self.kabiliyetleri) # buraya kadar olan kod çalıştığında çektı vermez.
        #kodların çıktı verebilmesi için mutlaka örneklenmeleri lazım.
Çalışan()
#  Çünkü self.kabiliyetleri bir sınıf niteliği değil, bir örnek niteliğidir. Örnek niteliklerine
#erişebilmek için de ilgili sınıfı mutlaka örneklememiz gerekir. Ayrıca sınıf niteliklerinin
#aksine, örnek niteliklerine sınıf adları üzerinden erişemeyiz. Yani self.kabiliyetleri adlı örnek
#niteliğine erişmeye yönelik şöyle bir girişim bizi hüsrana uğratacaktır:
# Çalışan.kabiliyetleri örneklendirme mekanizması gerekli
Çalışan().kabiliyetleri #Paranteze dikkat. sınıf niteliğine erişmek için sınıf adı kullanılır
# Örenklendirilen sınıfı bir örneğe atayalım:
ahmet=Çalışan() #Çalışan sınıfından bir nesne tanımlanıyor.
print(ahmet.kabiliyetleri) #örnek niteliğine ulaşmak için örnek adı kullanılır.
#  self kelimesi, yukarıdaki kodda yer alan ahmet kelimesini temsil ediyor. Yani
#ahmet.kabiliyetleri şeklinde bir kod yazabilmemizi sağlayan şey, __init__() fonksiyonu
#içinde belirttiğimiz self kelimesidir.
```

```python
class Çalışan():
    def __init__(self,isim):
        self.isim=isim

ahmet=Çalışan("Ahmet")
#Bu adımda şu olur:
#1.isim parametresi olarak "Ahmet" gönderilir.
#2.self.isim = isim şu hale gelir: self.isim = "Ahmet"
    #self, burada ahmet nesnesini temsil eder.
    #Yani artık ahmet nesnesinin bir isim özelliği vardır ve değeri "Ahmet"tir.
print(ahmet.isim)
```
