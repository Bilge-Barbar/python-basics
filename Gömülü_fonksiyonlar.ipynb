# sorted()
#Sıralama yapar.
#Hangi veri tipi ile giriş yapılırsa yapılsın çıktı olarak liste verir.
#Türkçe karakterleri doğru sıralayamaz.
#Birden fazla ögeli verilerde ilk ögeye göre sıralama yapar. Mesela ilk öge sayı ise sayıları sıralar.


```python
harfler="abcçdefgğhıijklmnoöprsştuüvyz"
çevrim={i: harfler.index(i) for i in harfler}
print(çevrim)
isimler=["ali","ayşe","murat","ceren","kemal","osman","şule"]
sorted(isimler) #yanlış sıraladı çünkü ikinci harfleri dikkate almadı.

# Doğru sıralama için:
def sırala(kelime):
    return([çevrim.get(kelime[i])for i in range (len(kelime))]) #önce ilk sonra iki sonra... bakar = .get(kelime[i])
print(*sorted(isimler,key=sırala),sep="\n")

#Başka bir örnek:

def sıralaa(liste):
    return liste[1] #ikinci sıradaki ögeleri yani yaşları baz alarak sıralar.

elemanlar=[("ali",21,"erzurum"),
            ("ayşe",12,"malatya"),
            ("murat",44,"şanlıurfa"),
            ("serkan",50,"iskenderun"),
            ("özge",55,"balıkesir")]
    
print(*sorted(elemanlar,key=sıralaa),sep="\n")
```
